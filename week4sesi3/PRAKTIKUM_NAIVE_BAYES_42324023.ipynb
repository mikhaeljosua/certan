```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
```

## 1. Memuat dataset Iris
Dataset Iris dimuat dari scikit-learn, lalu dikonversi menjadi DataFrame agar nama setiap fitur terlihat jelas.


```python
# Load dataset Iris dari scikit-learn
iris = load_iris()

# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]

print(X)
y
```

## 2. Label Encoding
Nama species diubah menjadi nilai numerik: setosa = 0, versicolor = 1, dan virginica = 2.


```python
from sklearn.preprocessing import LabelEncoder

lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

## 3. Membagi data latih dan data uji
Sebanyak 80% data digunakan untuk pelatihan dan 20% untuk pengujian. `random_state=0` membuat pembagian data konsisten dengan modul.


```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)

print("Jumlah data latih:", len(X_train))
print("Jumlah data uji:", len(X_test))
```

## 4. Melatih Gaussian Naïve Bayes
Fitur sepal length, sepal width, petal length, dan petal width berupa data kontinu, sehingga digunakan `GaussianNB`.


```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
result = NB_model.predict(X_test)
result
```

## 5. Mengukur akurasi
Prediksi data uji dibandingkan dengan label sebenarnya menggunakan `accuracy_score`.


```python
from sklearn.metrics import accuracy_score

print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

## 6. Prediksi input baru dari contoh modul
Prediksi pertama menggunakan list secara langsung. Hasilnya benar, tetapi scikit-learn dapat menampilkan peringatan karena input tidak memiliki nama fitur.


```python
# Prediksi inputan data baru
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Agar struktur input konsisten dengan data saat model dilatih, input diubah menjadi DataFrame dengan nama kolom yang sama.


```python
input_data = pd.DataFrame(
    np.array([[5.1, 3.5, 1.4, 0.2]]),
    columns=iris.feature_names
)
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

## Tugas: Prediksi `[7.3, 3.8, 4.6, 2.3]`
Data tugas dibuat sebagai DataFrame dengan urutan dan nama fitur yang sama seperti dataset Iris.


```python
data_tugas = pd.DataFrame(
    np.array([[7.3, 3.8, 4.6, 2.3]]),
    columns=iris.feature_names
)

hasil_tugas = NB_model.predict(data_tugas)
species_tugas = lab.inverse_transform(hasil_tugas)

print("Data baru:", data_tugas.iloc[0].tolist())
print("Kode kelas:", hasil_tugas[0])
print("Hasil prediksi species:", species_tugas[0])
```
