### 4. Ringkasan statistik dengan describe()

Untuk memahami data dari ringkasan statistik, DataFrame menyediakan fungsi `describe()`.

```python
df_noid.describe()
```

- **count** menunjukkan jumlah total data atau observasi dalam kolom tersebut
- **mean** adalah nilai rata-rata aritmetik dari seluruh data pada kolom
- **std** (standar deviasi) mengukur seberapa besar variasi atau penyebaran data dari nilai rata-rata
- **min** menyatakan data terkecil pada kolom
- **25%** menunjukkan kuartil pertama (Q1) dari data pada kolom
- **50%** menunjukkan titik tengah data (median) atau kuartil kedua (Q2) dari data pada kolom
- **75%** menyatakan kuartil ketiga (Q3) dari data pada kolom
- **max** menyatakan data maksimum atau terbesar dari data pada kolom

### 5. Mengambil rata-rata kolom tertentu

### 6. Korelasi antar data

Jika ingin mengetahui hubungan atau korelasi antar data, bisa dengan menggunakan fungsi `corr()`. Nilai korelasi berkisar dari -1 hingga 1.

- 1 = korelasi positif (ketika satu variabel naik, yang lain ikut naik)
- -1 = korelasi negatif (ketika satu variabel naik, yang lain turun)
- 0 = tidak ada korelasi

```python
df_noid[['temperature_celsius', 'humidity','air_quality_Ozone']].corr()
```

Dari data di atas, dapat dipahami bahwa:

- **Temperature vs Air Quality Ozone**: korelasi positif, berarti semakin naik suhu maka kadar ozon juga meningkat.
- **Humidity vs Air Quality Ozone**: korelasi negatif, berarti semakin naik kelembaban udara maka kadar ozon menurun.
- **Kesimpulan**: kenaikan suhu meningkatkan kadar ozon, sedangkan kenaikan kelembaban udara menurunkan kadar ozon - pola ini sama seperti contoh yang diberikan pada modul, meskipun angka pastinya bisa sedikit berbeda tergantung data.

### Tugas 2
Gunakan dataset [Customer Shopping Trends Dataset](https://www.kaggle.com/datasets/iamsouravbanerjee/customer-shopping-trends-dataset).

- Analisa mean, std, min, max, Q1, Q2, Q3 untuk data `Previous Purchases`
- Analisa korelasi `Age` vs `Purchase Amount (USD)` dan `Purchase Amount (USD)` vs `Review Rating`

```python
df2 = pd.read_csv("shopping_trends_updated.csv")
df2
```

#### Statistik deskriptif kolom `Previous Purchases`

```python
df2['Previous Purchases'].describe()
```

Nilai Q1, Q2 (median), dan Q3 secara eksplisit:

```python
q1 = df2['Previous Purchases'].quantile(0.25)
q2 = df2['Previous Purchases'].quantile(0.50)
q3 = df2['Previous Purchases'].quantile(0.75)
print(f"Q1 = {q1}")
print(f"Q2 (median) = {q2}")
print(f"Q3 = {q3}")
```

**Interpretasi:**

- **count** = 3900 data pelanggan.
- **mean** = 25,35, artinya rata-rata pelanggan sudah melakukan sekitar 25 kali pembelian sebelumnya.
- **std** = 14,45, menunjukkan variasi jumlah pembelian sebelumnya cukup besar antar pelanggan.
- **min** = 1 dan **max** = 50, rentang jumlah pembelian sebelumnya cukup lebar.
- **Q1** = 13, **Q2 (median)** = 25, **Q3** = 38, artinya separuh pelanggan memiliki riwayat pembelian sebelumnya antara 13 hingga 38 kali, dan distribusinya terlihat cukup merata/simetris karena median (25) hampir sama dengan mean (25,35).

#### Korelasi `Age` vs `Purchase Amount (USD)`

```python
df2[['Age', 'Purchase Amount (USD)']].corr()
```

#### Korelasi `Purchase Amount (USD)` vs `Review Rating`

```python
df2[['Purchase Amount (USD)', 'Review Rating']].corr()
```

**Interpretasi:**

- **Age vs Purchase Amount (USD)**: korelasi bernilai sekitar -0,01, sangat mendekati 0. Ini berarti usia pelanggan praktis tidak berhubungan dengan besar nilai pembelian yang mereka lakukan - pelanggan muda maupun tua cenderung membelanjakan jumlah yang serupa.
- **Purchase Amount (USD) vs Review Rating**: korelasi juga sangat lemah, sekitar 0,03. Artinya, besar kecilnya nominal pembelian tidak berhubungan secara linear dengan tinggi rendahnya rating ulasan yang diberikan pelanggan.

**Kesimpulan:** kedua pasangan variabel ini sama-sama menunjukkan hubungan linear yang sangat lemah (mendekati 0), sehingga `Age` dan `Review Rating` bukan prediktor yang kuat untuk `Purchase Amount (USD)` pada dataset ini.
