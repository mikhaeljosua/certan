```python
import numpy as np

# 1. Membuat matriks 5x5 dengan elemen random (bilangan bulat 1-100)
matriks = np.random.randint(1, 110, size=(5, 5))
print("Matriks 5x5 (elemen random):")
print(matriks)
print()

# 2. Menampilkan nilai maksimum, minimum, dan rata-rata
nilai_max = matriks.max()
nilai_min = matriks.min()
nilai_rata = matriks.mean()

print(f"Nilai Maksimum : {nilai_max}")
print(f"Nilai Minimum  : {nilai_min}")
print(f"Nilai Rata-rata: {nilai_rata:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Definisi fungsi persamaan
def persamaan_2d(x, y):
    return (x**6 + 3*x**4*y**2 - 3*x**4 + 3*x**2*y**4 - 
            x**2*y**3 - 6*x**2*y**2 + 3*x**2 + y**6 - 3*y**4 + 3*y**2 - 1)

# Membuat grid titik
x = np.linspace(-2, 2, 400)
y = np.linspace(-2, 2, 400)
X, Y = np.meshgrid(x, y)
Z = persamaan_2d(X, Y)

# Plot kontur untuk persamaan = 0
plt.figure(figsize=(8, 6))
contour = plt.contour(X, Y, Z, levels=[0], colors='blue', linewidths=2)
plt.grid(True, alpha=0.3)
plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')
plt.title('Grafik 2D: $x^6 + 3x^4y^2 - 3x^4 + 3x^2y^4 - x^2y^3 - 6x^2y^2 + 3x^2 + y^6 - 3y^4 + 3y^2 = 1$')
plt.axis('equal')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits import mplot3d

# Definisi fungsi 3D
def f(x, y):
    return np.exp(-np.abs(x)) * ((2 + x**2) / (5 * y**4))

# Membuat grid titik (hindari y = 0 karena pembagian dengan nol)
x = np.linspace(-3, 3, 50)
y = np.linspace(0.5, 3, 50)  # y mulai dari 0.5 untuk menghindari singularitas
X, Y = np.meshgrid(x, y)
Z = f(X, Y)

# Plot 3D
fig = plt.figure(figsize=(10, 7))
ax = plt.axes(projection='3d')

# Plot surface
surf = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none', alpha=0.8)

# Label dan judul
ax.set_xlabel('Sumbu X')
ax.set_ylabel('Sumbu Y')
ax.set_zlabel('Sumbu Z')
ax.set_title('Grafik 3D: $f(x,y) = e^{-|x|} \\cdot \\frac{2 + x^2}{5y^4}$')

# Menambahkan colorbar
fig.colorbar(surf, ax=ax, shrink=0.5, aspect=10)

plt.show()
```
